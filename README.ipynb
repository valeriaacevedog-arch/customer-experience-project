# __Vanguard Digital Interface Redesign: A/B Test Analysis__
## Project Overview

This project analyzes the results of a massive __A/B test__ conducted for Vanguard, a leading investment management company. Vanguard recently introduced a sleek, modern user interface (__Test group__) to replace their traditional client portal (__Control group__).

The main objective of this analysis is to determine if the new design actually improves the user experience, increases the process completion rate, and ensures that Vanguard's core demographic (older clients) is not left behind by the digital update.

### Data & Methodology
The analysis is based on user interaction logs, tracking each step clients took from the initial prompt (`start`) to the final confirmation (`confirm`).

To avoid skewed data caused by extreme outliers (e.g., users leaving a tab open for days), median values were used for time-based metrics instead of mathematical averages. Statistical significance was verified using Z-Tests for proportions and Welch's T-Tests for continuous variables.

### Metrics Tracked
To evaluate the success of the new design, the following six Key Performance Indicators (KPIs) were calculated and analyzed:

- KPI 1 - Completion Rate: The percentage of users who successfully completed the process (reached the confirm step) out of those who started.

- KPI 2 - Time per Step: The median time (in seconds) users spent on each individual screen, used to identify UX bottlenecks.

- KPI 3 - Error Rate: The percentage of users who navigated backward to a previous step, indicating confusion or hesitation.

- KPI 4 - Drop-off Rate per Step: The specific percentage of users who abandoned the process at each individual screen.

- KPI 5 - Total Time to Completion: The overall median time it took a user to travel from the start screen to the confirm screen.

- KPI 6 - Segmented Completion Rate by Age: The completion rate broken down by user age brackets to ensure demographic inclusivity.

#### Key Findings
Overall Success (KPI 1 & 5): The new design is a clear winner. The completion rate increased by 3.71 percentage points (from 65.58% in the Control group to 69.29% in the Test group). Furthermore, users navigated the new interface faster, with the median overall completion time dropping from 74.4 seconds to 68.5 seconds.

The "Step 1" Bottleneck (KPI 2, 3 & 4): Despite the overall success, the new design introduced friction at step_1. Users in the Test group spent more time, made more errors (navigating backward), and abandoned the process at a slightly higher rate during this specific step compared to the Control group. This is the primary area recommended for UX review.

Demographic Impact (KPI 6): The redesign is universally inclusive. The completion rate improved across all age brackets. Most importantly, clients aged 60 and older saw a 4.8-point increase in their success rate, proving the modern UI did not alienate the older demographic.
